# 🧪 Test de la Collecte de Données (ETL - Step 1)

Ce notebook permet de tester interactivement la classe `DataCollector` pour les 3 sources de données :
1. **Légifrance (API PISTE)** : Articles du Code du Travail (R4222).
2. **INRS (Web Scraping)** : Guides PDF et pages web sur l'aération.
3. **ARIA (Open Data)** : Base de données des accidents industriels.

---

## 1. Configuration de l'environnement
On ajoute le dossier racine au `sys.path` pour pouvoir importer le module `src`.

```python
import sys
import os
from pathlib import Path
import pandas as pd

# Ajout du dossier racine au path
project_root = Path("..").resolve()
sys.path.append(str(project_root))

print(f"Project Root: {project_root}")

# Import du collecteur
from src.etl.collect import DataCollector
from src.config import get_settings

# Initialisation
collector = DataCollector()
settings = get_settings()

print("✅ Environnement chargé.")
```

## 2. Test Légifrance (API PISTE)
Cette étape nécessite des clés valides dans `.env` (`LEGIFRANCE_CLIENT_ID` et `LEGIFRANCE_CLIENT_SECRET`).

> **Note :** Si l'authentification échoue (401/400), vérifiez vos accès sur https://piste.gouv.fr.

```python
# Test de connexion et récupération
df_legifrance = collector.collect_legifrance()

if not df_legifrance.empty:
    print(f"\n✅ {len(df_legifrance)} articles récupérés.")
    display(df_legifrance.head())
else:
    print("\n⚠️ Aucune donnée récupérée (Vérifiez les logs/clés API).")
```

## 3. Test INRS (Web Scraping)
Récupère les titres des guides depuis le site de l'INRS.

```python
df_inrs = collector.collect_inrs()

if not df_inrs.empty:
    print(f"\n✅ {len(df_inrs)} guides trouvés.")
    display(df_inrs.head())
else:
    print("\n⚠️ Problème de scraping INRS.")
```

## 4. Test ARIA (Open Data)
Charge le fichier CSV local (`data/raw/aria/aria.csv`).

```python
df_aria = collector.collect_aria()

if not df_aria.empty:
    print(f"\n✅ {len(df_aria)} accidents chargés.")
    display(df_aria.head())
    print("\nColonnes disponibles :", df_aria.columns.tolist())
else:
    print("\n⚠️ Fichier ARIA non trouvé ou vide.")
```

## 5. Vérification des fichiers de sortie
On vérifie que les fichiers CSV ont bien été générés dans `data/raw/` ou `data/processed/`.

```python
output_dir = Path("../data/raw")
files = list(output_dir.glob("*.csv"))

print(f"📂 Fichiers dans {output_dir} :")
for f in files:
    print(f" - {f.name} ({f.stat().st_size / 1024:.2f} KB)")
```
